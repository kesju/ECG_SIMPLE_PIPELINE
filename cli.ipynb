# ECG Denoising Pipeline Demo
This notebook shows how to run the ECG denoising pipeline and visualise the resulting annotations.

```python
from pathlib import Path
from ecg_pipeline import (
    ECGDenoisingPipeline,
    load_config,
    load_ecg_from_npy,
    load_segments_from_json,
    plot_ecg_with_annotations,
)

config = load_config(Path('config/pipeline_config.json'))
ecg_orig = load_ecg_from_npy(Path('data/ecg_orig.npy'))
gaps_segments = load_segments_from_json(Path('data/gaps_indexes.json'))
pipeline = ECGDenoisingPipeline(config)
result = pipeline.run(ecg_orig, gaps_segments)

len_orig = result.ecg_orig.shape[0]
len_start = result.ecg_start.shape[0]
len_final = result.ecg_final.shape[0]
print(f'Original samples: {len_orig}')
print(f'After gap removal: {len_start}')
print(f'Final samples: {len_final}')
```

```python
annotations = {
    'outliers': result.projected_outliers,
    'rdropouts': result.projected_rdropouts,
}
plot_ecg_with_annotations(
    result.ecg_start,
    sample_rate=config.plotting.sample_rate,
    annotations=annotations,
    total_points=config.plotting.total_points,
    title='ECG start with anomalies',
    show=True,
)
```
